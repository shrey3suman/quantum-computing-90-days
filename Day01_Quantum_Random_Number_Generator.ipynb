```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

qc = QuantumCircuit(1,1)
qc.h(0)
qc.measure(0,0)

simulator = AerSimulator()
job = simulator.run(qc, shots=10, memory=True)
result = job.result()

random_bits = result.get_memory()

print ("Your 10 Quantum Random Bits:", random_bits)
print ("Summary Counts:", result.get_counts())
```
